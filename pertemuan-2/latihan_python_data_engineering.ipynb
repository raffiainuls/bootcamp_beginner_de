# Latihan Python untuk Data Engineering

Notebook ini mempraktikkan pembuatan data sederhana, pembacaan CSV/JSON, penggunaan NumPy, transformasi, dan penyimpanan output. Setiap baris kode diberi komentar agar mudah diikuti.

## Tujuan Latihan

- Membuat dan membaca data tabular dengan Pandas.
- Membaca file CSV dan JSON.
- Menggunakan NumPy untuk operasi array dan numerik.
- Memeriksa struktur data.
- Rename kolom dan filter baris.
- Menyimpan output transformasi ke CSV.

```python
# Mengimpor NumPy untuk operasi array dan perhitungan angka.
import numpy as np
# Mengimpor Pandas untuk membuat dan mengolah DataFrame.
import pandas as pd
# Mengimpor Path untuk bekerja dengan lokasi file dan folder.
from pathlib import Path
```

## 1. Membuat DataFrame Sederhana

```python
# Membuat DataFrame dari dictionary berisi nama kolom dan nilai.
data = pd.DataFrame({
    # Membuat kolom ID transaksi.
    'transaction_id': ['TRX-001', 'TRX-002', 'TRX-003', 'TRX-004', 'TRX-005'],
    # Membuat kolom nama pelanggan.
    'customer': ['Andi', 'Budi', 'Citra', 'Dewi', 'Eka'],
    # Membuat kolom jumlah transaksi.
    'amount': [150000, 75000, 225000, 100000, 350000]
})
data
```

## 2. Membaca File CSV dan JSON

```python
# Menentukan folder tempat file CSV dan JSON berada.
data_dir = Path('data')
# Membaca file CSV menjadi DataFrame.
data_csv = pd.read_csv(data_dir / 'transaksi.csv')
# Membaca file JSON menjadi DataFrame.
data_json = pd.read_json(data_dir / 'transaksi.json')

# Menampilkan label data CSV.
print('Data dari CSV:')
# Menampilkan isi data CSV.
display(data_csv)

# Menampilkan label data JSON.
print('Data dari JSON:')
# Menampilkan isi data JSON.
display(data_json)
```

```python
# Menampilkan lima baris pertama dari data.
data.head()
# Menampilkan nama kolom, jumlah baris, dan tipe data.
data.info()
# Menampilkan ringkasan statistik kolom angka.
data.describe()
```

## 3. Praktik NumPy

Kolom `amount` diubah menjadi array NumPy untuk menghitung total dan rata-rata.

```python
# Mengubah kolom amount menjadi array NumPy.
nilai_transaksi = np.array(data['amount'])
# Menghitung total seluruh transaksi.
total_transaksi = np.sum(nilai_transaksi)
# Menghitung rata-rata seluruh transaksi.
rata_rata_transaksi = np.mean(nilai_transaksi)
# Menampilkan total transaksi.
print(f'Total transaksi: Rp{total_transaksi:,.0f}')
# Menampilkan rata-rata transaksi.
print(f'Rata-rata transaksi: Rp{rata_rata_transaksi:,.0f}')
```

## 4. Transformasi Data Sederhana

```python
# Mengganti nama kolom amount menjadi total_transaksi.
hasil = data.rename(columns={'amount': 'total_transaksi'})
# Memilih transaksi dengan nilai lebih besar dari 100000.
hasil = hasil[hasil['total_transaksi'] > 100000].copy()
# Mengubah kolom hasil filter menjadi array NumPy.
nilai_terfilter = hasil['total_transaksi'].to_numpy()
# Menghitung rata-rata transaksi yang lolos filter.
rata_rata_terfilter = np.mean(nilai_terfilter)
# Menghitung total transaksi yang lolos filter.
total_terfilter = np.sum(nilai_terfilter)
# Menampilkan rata-rata transaksi yang lolos filter.
print(f'Rata-rata transaksi yang lolos filter: Rp{rata_rata_terfilter:,.0f}')
# Menampilkan total transaksi yang lolos filter.
print(f'Total transaksi yang lolos filter: Rp{total_terfilter:,.0f}')
# Menampilkan data hasil transformasi.
hasil
```

```python
# Menentukan folder output untuk menyimpan hasil.
output_dir = Path('output')
# Membuat folder output jika belum ada.
output_dir.mkdir(exist_ok=True)
# Menentukan nama file hasil transformasi.
output_file = output_dir / 'transaksi_terfilter.csv'
# Menyimpan hasil ke CSV tanpa index tambahan.
hasil.to_csv(output_file, index=False)
# Menampilkan lokasi file hasil.
print(f'Hasil tersimpan di: {output_file}')
```
