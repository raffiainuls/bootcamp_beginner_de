# Latihan Dasar Python untuk Pemula

Contoh use case: mencatat transaksi sederhana dan mengecek transaksi sebelum diproses oleh pipeline data. Latihan ini sengaja tidak menggunakan Pandas atau NumPy. Setiap baris kode diberi komentar agar mudah diikuti.

## 1. Variabel dan Tipe Data

Variabel digunakan untuk menyimpan informasi yang dibutuhkan program.

```python
# Menyimpan nama pipeline ke dalam variabel string.
nama_pipeline = 'Pipeline Transaksi'
# Menyimpan jumlah transaksi ke dalam variabel integer.
jumlah_transaksi = 3
# Menyimpan status pipeline ke dalam variabel boolean.
pipeline_aktif = True

# Menampilkan nama pipeline ke output notebook.
print('Nama pipeline:', nama_pipeline)
# Menampilkan jumlah transaksi ke output notebook.
print('Jumlah transaksi:', jumlah_transaksi)
# Menampilkan status pipeline ke output notebook.
print('Pipeline aktif:', pipeline_aktif)
```

## 2. List dan Perulangan

List dapat menyimpan beberapa nilai. `for` dapat digunakan untuk membaca setiap nilai satu per satu.

```python
# Membuat list yang berisi beberapa nama pelanggan.
nama_pelanggan = ['Andi', 'Budi', 'Citra']

# Mengambil nama pelanggan satu per satu dari dalam list.
for nama in nama_pelanggan:
    # Menampilkan nama pelanggan yang sedang diproses.
    print('-', nama)
```

## 3. Dictionary sebagai Satu Record

Dictionary menyimpan data dalam pasangan nama kolom dan nilai. Dalam data engineering, dictionary dapat digunakan untuk mewakili satu record.

```python
# Membuat dictionary untuk menyimpan satu transaksi.
transaksi = {
    # Menyimpan ID transaksi.
    'id': 'TRX-001',
    # Menyimpan nama pelanggan.
    'pelanggan': 'Andi',
    # Menyimpan jumlah transaksi dalam bentuk integer.
    'jumlah': 150000,
    # Menyimpan status pembayaran transaksi.
    'status': 'paid',
}

# Menampilkan seluruh isi dictionary transaksi.
print('Satu transaksi:', transaksi)
# Mengambil dan menampilkan nilai dari key pelanggan.
print('Nama pelanggan:', transaksi['pelanggan'])
```

## 4. Percabangan `if`

`if` digunakan untuk membuat keputusan, misalnya memeriksa status transaksi.

```python
# Memeriksa apakah status transaksi sama dengan paid.
if transaksi['status'] == 'paid':
    # Perintah ini dijalankan jika kondisi benar.
    print('Transaksi sudah dibayar')
else:
    # Perintah ini dijalankan jika kondisi salah.
    print('Transaksi belum dibayar')
```

## 5. Function

Function membantu kita menggunakan kembali aturan yang sama.

```python
# Membuat function dengan parameter bernama data.
def cek_transaksi(data):
    # Memeriksa status transaksi yang diterima function.
    if data['status'] == 'paid':
        # Mengembalikan teks valid jika transaksi sudah dibayar.
        return 'valid'
    # Mengembalikan teks lain jika transaksi belum dibayar.
    return 'perlu diperiksa'

# Memanggil function dengan data transaksi sebagai argumen.
print('Hasil pengecekan:', cek_transaksi(transaksi))
```
